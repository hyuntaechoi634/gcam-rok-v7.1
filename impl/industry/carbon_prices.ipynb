# Carbon Prices under the Korean Emissions Trading Scheme

```python
tco2_2022_2022krw = 22370
tco2_2021_2021krw = 23149
tco2_2020_2020krw = 30411
tco2_2019_2019krw = 28440
tco2_2018_2018krw = 22120
tco2_2017_2017krw = 20951
tco2_2016_2016krw = 17056

tc_2022_1990usd = tco2_2022_2022krw * krw_to_usd(2022) * gcam_deflator(1, 2022, 1990) * (44/12)
tc_2021_1990usd = tco2_2021_2021krw * krw_to_usd(2021) * gcam_deflator(1, 2021, 1990) * (44/12)
tc_2020_1990usd = tco2_2020_2020krw * krw_to_usd(2020) * gcam_deflator(1, 2020, 1990) * (44/12)
tc_2019_1990usd = tco2_2019_2019krw * krw_to_usd(2019) * gcam_deflator(1, 2019, 1990) * (44/12)
tc_2018_1990usd = tco2_2018_2018krw * krw_to_usd(2018) * gcam_deflator(1, 2018, 1990) * (44/12)
tc_2017_1990usd = tco2_2017_2017krw * krw_to_usd(2017) * gcam_deflator(1, 2017, 1990) * (44/12)
tc_2016_1990usd = tco2_2016_2016krw * krw_to_usd(2016) * gcam_deflator(1, 2016, 1990) * (44/12)

volume_2022 = 39140
volume_2021 = 54716
volume_2020 = 44013
volume_2019 = 38084
volume_2018 = 47507
volume_2017 = 26264
volume_2016 = 11969
volumn_tot = volume_2022 + volume_2021 + volume_2020 + volume_2019 + volume_2018 + volume_2017 + volume_2016

weight_2022 = volume_2022 / volumn_tot
weight_2021 = volume_2021 / volumn_tot
weight_2020 = volume_2020 / volumn_tot
weight_2019 = volume_2019 / volumn_tot
weight_2018 = volume_2018 / volumn_tot
weight_2017 = volume_2017 / volumn_tot
weight_2016 = volume_2016 / volumn_tot
```

```python
carbon_price_cp_tc_1990usd = tc_2022_1990usd * weight_2022 + tc_2021_1990usd * weight_2021 + tc_2020_1990usd * weight_2020 + tc_2019_1990usd * weight_2019 + tc_2018_1990usd * weight_2018 + tc_2017_1990usd * weight_2017 + tc_2016_1990usd * weight_2016
carbon_price_cp_tc_1990usd
```

```python
carbon_price_cp = {
    2020: tc_2020_1990usd,
    2025: carbon_price_cp_tc_1990usd,
    2030: carbon_price_cp_tc_1990usd * (1.045 ** 5),
    2035: carbon_price_cp_tc_1990usd * (1.045 ** 10),
}
carbon_price_cp
```

```python
carbon_price_ep = {
    2020: tc_2020_1990usd,
    2025: carbon_price_cp_tc_1990usd,
    2030: carbon_price_ep_tc_1990usd * (1.045 ** 5),
    2035: carbon_price_ep_tc_1990usd * (1.045 ** 10),
}
carbon_price_ep
```
