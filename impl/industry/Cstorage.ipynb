```python
import pandas as pd

dfDef = pd.read_csv("../resources/gdpdef.csv")
dfDef.head()
dfDef['Year'] = dfDef['date'].str.split('-').str[0].astype(int)
arrDef = dfDef.set_index('Year')['gdpdef']
arrDef[2024] = (139.71495 / 136.41462) * arrDef[2023]
def gcam_deflator(value, from_year=2023, to_year=1975):
    return value * arrDef[to_year] / arrDef[from_year]

dfExc = pd.read_csv("../resources/DEXKOUS.csv")
dfExc['date'] = pd.to_datetime(dfExc['observation_date'])
dfExcAnnual = dfExc.groupby(dfExc['date'].dt.year)['DEXKOUS'].mean()

def krw_to_usd(year):
    return 1 / dfExcAnnual[year]
```

# Storage Subsidy for CCUS

The Korean government projects the cost of CO₂ storage under CCUS to reach approximately USD 50 per tCO₂ (constant 2023 US dollars) by 2030, declining further to USD 45 per tCO₂ by 2050 as learning effects and scale economies materialize. However, the Act on the Capture, Transportation, Storage, and Utilization of Carbon Dioxide (CCUS Act), while providing a legal basis for financial, administrative, and technical support for CCS and CCUS projects, does not stipulate any explicit subsidy level per tonne of CO₂ captured and stored.

Accordingly, under the Current Policies scenario, this study assumes no explicit subsidy for CCS storage, reflecting the absence of a clearly defined and committed support scheme in current Korean policy. Under this scenario, CCS deployment in GCAM is therefore constrained by the model’s default offshore carbon storage cost assumptions.

Under the Enhanced Ambition scenario, by contrast, policy support is assumed to be introduced to align the effective cost of CO₂ storage with the government’s projected cost trajectory. Specifically, subsidies are calibrated such that the net storage cost faced by emitters converges toward USD 50 per tCO₂ by 2030 and declines to USD 45 per tCO₂ by 2050 (constant 2023 US dollars). While this support level is not derived from an announced Korean policy instrument, it is broadly consistent with the magnitude of CCS incentives observed in leading jurisdictions, notably the USD 84 per tCO₂ tax credit for dedicated geological storage under the U.S. Inflation Reduction Act’s enhanced 45Q provision. In this study, the 45Q benchmark is used solely to assess the policy plausibility of the assumed support level, rather than as a direct analogue for Korean policy design.
