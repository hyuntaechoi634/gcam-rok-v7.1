# Chemical

This section documents how mitigation options in the Korean petrochemical industry are represented in the model, distinguishing between *current policy* and *enhanced policy* scenarios and explaining how these assumptions are translated into parameter changes in GCAM–ROK.

---

## 1. Background: feedstock and fuel substitution in the petrochemical industry

The Korean petrochemical industry has identified **feedstock substitution** (e.g., recycled or alternative feedstocks replacing fossil naphtha) and **fuel substitution** (e.g., electrification or hydrogen use in naphtha cracking furnaces) as key long-term decarbonization pathways (Lee, 2024). However, as of today, **there is no policy- or technology-level commitment that can be regarded as firmly established or binding**, either in terms of mandatory substitution targets or large-scale commercial deployment.

As a result, while these pathways are frequently discussed in roadmaps and scenario studies, they remain **aspirational rather than realized** under currently implemented policies.

---

## 2. Treatment under the *Current Policies* scenario

Given the absence of clearly defined and committed policy instruments or commercially mature deployment at scale, the *Current Policy* scenario assumes that **no additional mitigation measures are explicitly implemented in the petrochemical sector** beyond existing baseline efficiencies.

Accordingly:
- No feedstock substitution is imposed beyond the default model structure.
- No furnace electrification or hydrogen switching is assumed.
- Parameters governing petrochemical production in GCAM–ROK are left unchanged from their reference configuration.

This reflects a conservative interpretation in which ongoing discussions and pilot projects do not yet translate into sector-wide emission reductions.

---

## 3. Treatment under the *Enhanced Ambition* scenario

By contrast, the *Enhanced Policy* scenario represents a **feasible but highly ambitious decarbonization pathway**, informed by scenario analyses in Nextgroup (2024). Under this scenario, it is assumed that Korea actively pursues both feedstock and fuel substitution in the petrochemical industry through coordinated policy support, infrastructure investment, and technology deployment.

To ensure internal consistency, these assumptions are translated into **explicit parameter adjustments in the chemical sector of GCAM–ROK**, as detailed below.

---

## 4. Feedstock substitution: refined liquids input coefficient

### Concept
Feedstock substitution is interpreted as the replacement of fossil naphtha with alternative feedstocks, such as recycled naphtha derived from chemical recycling of waste plastics or other low-carbon feedstocks. In GCAM, this effect is represented by a **reduction in the refined liquids input coefficient** in the chemical feedstocks sector.

### Parameter adjustment
Let $a_t$ denote the refined liquids input coefficient in year $t$, and let $s_t$ denote the share of fossil feedstock substituted. The adjusted coefficient is given by:

$$
a_t^{\text{new}} = a_t \times (1 - s_t)
$$

### Assumed substitution rates
Based on the scenario trajectories reported in Nextgroup (2024), the following substitution rates are assumed:

- **2030**: $s_{2030} = 0.05$  
- **2035**: $s_{2035} = 0.135$

This corresponds to a **5% reduction in fossil feedstock use by 2030** and a **13.5% reduction by 2035**.

---

## 5. Fuel substitution: adjustment of the remove fraction

### Concept
Fuel substitution refers to the partial replacement of combustion-based heat supply in naphtha cracking furnaces with electricity or hydrogen. In this analysis, the GCAM *remove fraction* ($rf$) is interpreted such that:

- $1 - rf$ represents the share of refined liquids–derived carbon that is converted into off-gas and subsequently combusted in furnaces, resulting in CO$_2$ emissions.
- Electrification or hydrogen use reduces this combustion-related emission component.

Under this interpretation, fuel substitution is approximated by an **increase in $rf$**.

### Baseline configuration
The default GCAM configuration assumes:

$$
rf_0 = 0.75
$$

implying a baseline combustion-related emission fraction of:

$$
e_0 = 1 - rf_0 = 0.25
$$

### Linking heat substitution to $rf$
Let $p_t$ denote the share of total furnace heat demand replaced by electricity or hydrogen in year $t$. The remaining combustion-related emission fraction is:

$$
e_t = e_0 \times (1 - p_t)
$$

and the updated remove fraction is:

$$
rf_t = 1 - (1 - rf_0)(1 - p_t)
$$

### Assumed substitution rates
Consistent with a highly ambitious but feasible transition:

- **2030**: $p_{2030} = 0.20$  
- **2035**: $p_{2035} = 0.28$

### Resulting values
- **2030**:
  $$
  rf_{2030} = 0.80
  $$
- **2035**:
  $$
  rf_{2035} = 0.82
  $$

---

## 6. Summary

Under the *Enhanced Policy* scenario, the Korean petrochemical industry is assumed to undergo both feedstock and fuel substitution, guided by ambitious scenario pathways rather than currently binding policies. These assumptions are implemented in GCAM–ROK through:

- A **reduction in the refined liquids input coefficient** to represent fossil feedstock substitution, and
- An **increase in the remove fraction** to represent reduced combustion-related emissions from furnace electrification or hydrogen use.

Together, these adjustments provide a stylized but internally consistent representation of a highly ambitious decarbonization pathway for the petrochemical sector.
