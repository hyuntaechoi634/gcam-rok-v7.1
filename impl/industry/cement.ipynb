# Cement

## Current Policies

The expansion of blended cement is implemented conservatively by adjusting input coefficients of the existing Portland cement technology in GCAM, without introducing new inputs or modifying technology shares. Cement composition is specified exogenously as

$$
(s_{\mathrm{cl}}, s_{\mathrm{cc}}, s_{\mathrm{ls}}, s_{\mathrm{g}})
=
\begin{cases}
(0.85,\,0.08,\,0.02,\,0.05) & \text{in 2030},\\
(0.825,\,0.08,\,0.045,\,0.05) & \text{in 2035},
\end{cases}
$$

with the baseline Portland cement interpreted as $(0.95,0,0,0.05)$.

Let $L^{\mathrm{base}}=1.5$ denote the baseline limestone coefficient. Total limestone demand is calculated as

$$
L_t
=
\left(\frac{L^{\mathrm{base}}}{0.95}\right)s_{\mathrm{cl},t}
+ s_{\mathrm{ls},t},
$$

where the first term represents clinker-related limestone and the second term represents non-carbonate limestone filler.

Thermal energy requirements are adjusted using an energy-balance formulation:

$$
m^{\mathrm{heat}}_t
=
\frac{
s_{\mathrm{cl},t}E_{\mathrm{cl}} + s_{\mathrm{cc},t}E_{\mathrm{cc}}
}{
0.95\,E_{\mathrm{cl}}
},
\qquad
H_t = H^{\mathrm{base}}_t \, m^{\mathrm{heat}}_t,
$$

where $E_{\mathrm{cl}}$ and $E_{\mathrm{cc}}$ denote specific energy requirements for clinker calcination and calcined clay production, respectively.

Electricity demand is adjusted conservatively as

$$
m^{\mathrm{elec}}_t
=
1 + \kappa\big[(1-s_{\mathrm{cl},t})-0.05\big],
\qquad
E_t = E^{\mathrm{base}} \, m^{\mathrm{elec}}_t,
$$

where $\kappa$ is a small scaling parameter.

This formulation yields a lower-bound estimate of mitigation impacts, consistent with gradual adoption of blended cement technologies under current policy constraints.

## Enhanced Policy

Under the enhanced policy scenario, the blended cement formulation is assumed to immediately converge to an LC3-type composition and remain constant over time. Accordingly, cement composition is specified as

$$
(s_{\mathrm{cl}}, s_{\mathrm{cc}}, s_{\mathrm{ls}}, s_{\mathrm{g}})
=
(0.50,\,0.30,\,0.15,\,0.05)
\quad \text{for } t\in\{2030,2035\}.
$$

All other modeling conventions follow the current policy case, including the retention of the existing Portland cement technology, the absence of additional material inputs, and unchanged technology share-weights.

With the LC3 composition fixed, the limestone input coefficient becomes time-invariant and is given by

$$
L_t
=
\left(\frac{L^{\mathrm{base}}}{0.95}\right)s_{\mathrm{cl}} + s_{\mathrm{ls}}
=
0.94,
\qquad t\in\{2030,2035\}.
$$

Thermal energy requirements are adjusted using the same energy-balance formulation as in the current policy case. With $E_{\mathrm{cl}}$ and $E_{\mathrm{cc}}$ denoting specific energy requirements for clinker calcination and calcined clay production, respectively, the resulting heat multiplier is

$$
m^{\mathrm{heat}}
=
\frac{
s_{\mathrm{cl}}E_{\mathrm{cl}} + s_{\mathrm{cc}}E_{\mathrm{cc}}
}{
0.95\,E_{\mathrm{cl}}
}
=
0.72.
$$

The adjusted process heat coefficients are therefore

$$
H_{2030}=0.0037778\times0.72,
\qquad
H_{2035}=0.0037222\times0.72.
$$

Electricity demand is adjusted using the same proportional rule as in the current policy case. With $\kappa=0.2$, the electricity multiplier under LC3 adoption is

$$
m^{\mathrm{elec}}
=
1 + \kappa\big[(1-s_{\mathrm{cl}})-0.05\big]
=
1.09,
$$

implying identical electricity input coefficients in 2030 and 2035.

Relative to the current policy scenario, the enhanced policy represents a discrete shift to a mature low-clinker cement formulation from 2030 onward, yielding substantially larger reductions in clinker use and process heat demand while maintaining consistency with the existing GCAM technology structure.
