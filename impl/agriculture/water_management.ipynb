# Rice Cultivation Water Management Policy

The Ministry of Agriculture, Food and Rural Affairs plans to subsidize:  
- **ID (Intermittent Drainage, 간단관개)** – Target adoption share: **61.1%**  
- **CF+WS (Continuous Flooding + Water-Saving, 논물얕게대기)** – Target adoption share: **10%**  

According to Kim et al. (2013), both ID and CF+WS reduce CH₄ emissions and increase N₂O emissions compared to continuous flooding (CF), resulting in a net decrease in GWP.  

- **Current Policy scenario**: Adoption shares as above (ID: 61.1%, CF+WS: 10%).  
- **Enhanced Ambition scenario**: CF+WS share increases to **20%** by 2035.  

GWP reduction rates are applied to the default GCAM emissions coefficients to ensure model consistency.

Below are calculations for implementing the policy.

```python
# Create the DataFrame
data = {
    'Practice': ['WS', 'CF+WS', 'ID', 'CF (Baseline)'],
    'CH4_Emissions_kg_per_ha': [60.8, 81.6, 194.9, 263.0],
    'N2O_Emissions_kg_per_ha': [0.019, 0.014, 0.009, 0.003],
    'CH4_Reduction_vs_CF_%': [-76.9, -69.0, -25.9, 0],
    'N2O_Increase_vs_CF_%': [533, 367, 200, 0],
}

df = pd.DataFrame(data)
df['GWP'] = [1.32, 1.85, 4.51, 6.02]
df['GWP']/= 6.02
df['GWP'] *= 100
df
```
