# Low-Methane Feed Policy

South Korea plans to develop low-protein and low-methane feed to reduce enteric fermentation emissions from livestock.  
By 2030, the plan targets a reduction of **0.751 MtCO₂-eq** out of the projected **4.758 MtCO₂-eq** emissions from enteric fermentation.
This is 15.8% reduction in enteric fermentation emissions.
However, specific policy instruments for implementation have not yet been established.  

Therefore, this measure is included only in the *Enhanced Ambition* scenario.
