# Performance per Vehicle by Mode

# Car (pass-km/veh)

* daily milage (km / veh / day) : 44.3 [TMACS, Other Fuel](https://tmacs.kotsa.or.kr/web/TG/TG200/TG2200/Tg1700_02.jsp?mid=S3080#)
* lifespan (year) : 15.3 [KADRA](http://www.tbnews.co.kr/news/view.php?idx=1784)
* load factor (pass/veh): 1.26 [KTDB](https://www.ktdb.go.kr/common/downloadFile.do?fileNm=/PassengerTripSurvey/2009_press01.pdf&dnFileNm=%EC%84%9C%EC%9A%B8%EC%8B%9C%20%EC%B6%9C%EA%B7%BC%EC%8B%9C%EA%B0%84%EB%8C%80%20%EB%82%98%ED%99%80%EB%A1%9C%20%EC%B0%A8%EB%9F%89%EB%B9%84%EC%9C%A8%2086.3%ED%8D%BC%EC%84%BC%ED%8A%B8.pdf)

$\text{performance (pass-km/veh)}=\text{daily milage (km/veh/day)} * 365 \text{(day)} * \text{lifespan (year)} * \text{load factor (pass/veh)}$

```python
daily_milage_per_veh = 44.3
days_in_a_year = 365
lifespan = 15.3
load_factor = 1.26
discount_rate = 0.045

annual_service = daily_milage_per_veh * days_in_a_year * load_factor
print(annual_service)

pass_km_per_veh_disc = discounted_lifetime_service(
    annual_service, lifespan, discount_rate
)

pass_km_per_veh_disc
```
