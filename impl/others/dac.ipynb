# Direct Air Capture

In the *2050 Carbon Neutrality Scenario B*, direct air capture (DAC) is planned to reduce 7.4 MtCO₂-eq of greenhouse gas emissions.  
In contrast, *Scenario A* does not include any reductions from DAC.  
To clearly differentiate between the two, the *Current Policies* scenario assumes no DAC deployment.  
In the *Enhanced Ambition* scenario, DAC deployment is assumed to start at 0 MtCO₂ in 2025 and increase linearly to 14.8 MtCO₂ by 2050.

````python
for year in range(2030, 2055, 5):
    const_mtco2e = 7.4 * 2 / 5 * (year-2025)/5
    const_mtc = const_mtco2e / 3.66667
    print(year, const_mtco2e, const_mtc)
````
