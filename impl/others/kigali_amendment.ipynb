# Kigali Amendment

South Korea has ratified the *Kigali Amendment* to the Montreal Protocol, committing to the phasedown of hydrofluorocarbons (HFCs).  
However, policy instruments to achieve the targeted HFC reductions remain insufficient.  

In the *Current Policy* scenario, we assume that no additional measures are introduced beyond existing domestic regulations to fulfill the Kigali Amendment.  
In the *Enhanced Ambition* scenario, we assume the introduction of an HFC constraint which assumes the linear pathwy to an 80% reduction of HFCs by 2045 compared to average 2020-2022 level, which is 30.4 MtCO2e according to GIR (2025)

```python
emiss_base = 30.4
emiss2045 = 30.4 * 0.2
for year in range(2030, 2041, 5):
    emiss = emiss_base + (emiss2045 - emiss_base) * (year - 2025) / (2045 - 2025)
    print(year, emiss)
print(2045, emiss2045)
```
