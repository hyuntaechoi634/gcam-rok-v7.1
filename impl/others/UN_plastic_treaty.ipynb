# UN Plastic Treaty Policy

The *UN Plastic Treaty* commits signatory countries to reduce plastic waste by **25% by 2040**.  
We assume that South Korea ratifies this treaty and implements measures to achieve the target reduction.  

Emission reduction impacts are estimated using the **WARM v16** model developed by the U.S. Environmental Protection Agency (EPA).  

- **Effect**:  
  - 0.777 MtCO₂-eq reduction by 2030  
  - 1.554 MtCO₂-eq reduction by 2035  
  - 2.331 MtCO₂-eq reduction by 2040
